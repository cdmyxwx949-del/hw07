```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import random

# ====================== 固定随机种子（适配TF2.x） ======================
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

# ====================== 【关键】适配你的路径 ======================
BASE_DIR = "/kaggle/input/datasets/paultimothymooney/chest-xray-pneumonia/chest_xray/"
TRAIN_DIR = os.path.join(BASE_DIR, "train")
TEST_DIR = os.path.join(BASE_DIR, "test")

# 路径检查
print(f"✅ TRAIN_DIR 路径：{TRAIN_DIR}")
print(f"✅ TRAIN_DIR 是否存在：{os.path.exists(TRAIN_DIR)}")
print(f"✅ TEST_DIR 是否存在：{os.path.exists(TEST_DIR)}")

# ====================== 超参数设置 ======================
IMG_SIZE = 150
BATCH_SIZE = 32
EPOCHS = 15
LR = 1e-4

# ====================== 数据增强 + 8:2 划分验证集（符合作业要求） ======================
# 训练集增强（防止过拟合）
datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=10,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    validation_split=0.2  # 从train里分20%做验证集，不用原始val文件夹
)

# 训练集
train_generator = datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="training",
    shuffle=True
)

# 验证集（从train划分，而非原始val）
val_generator = datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="validation",
    shuffle=True
)

# 测试集
test_datagen = ImageDataGenerator(rescale=1./255)
test_generator = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    shuffle=False
)

# ====================== 构建CNN模型 ======================
model = Sequential([
    Conv2D(32, (3,3), activation="relu", input_shape=(IMG_SIZE, IMG_SIZE, 3)),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    
    Conv2D(64, (3,3), activation="relu"),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    
    Conv2D(128, (3,3), activation="relu"),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    
    Conv2D(128, (3,3), activation="relu"),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    
    Flatten(),
    Dropout(0.5),
    Dense(256, activation="relu"),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer=Adam(learning_rate=LR),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 模型结构打印
model.summary()

# ====================== 训练模型 ======================
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    verbose=1
)

# ====================== 绘制训练曲线 ======================
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
plt.plot(history.history["accuracy"], label="Train Acc")
plt.plot(history.history["val_accuracy"], label="Val Acc")
plt.title("Accuracy Curve")
plt.legend()

plt.subplot(1,2,2)
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Val Loss")
plt.title("Loss Curve")
plt.legend()

plt.tight_layout()
plt.savefig("/kaggle/working/train_curve.png", dpi=300)
plt.show()

# ====================== 测试集评估（四项核心指标） ======================
y_true = test_generator.classes
y_pred = model.predict(test_generator)
y_pred = (y_pred > 0.5).astype(int).flatten()

acc = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print("\n===== 测试集最终指标 =====")
print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1 Score:  {f1:.4f}")

print("\n分类报告：")
print(classification_report(y_true, y_pred, target_names=["Normal", "Pneumonia"]))

# ====================== 绘制混淆矩阵 ======================
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", 
            xticklabels=["Normal", "Pneumonia"], 
            yticklabels=["Normal", "Pneumonia"])
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.savefig("/kaggle/working/confusion_matrix.png", dpi=300)
plt.show()
```

```python
# ==========================
# 肺炎三分类：NORMAL / VIRAL / BACTERIAL
# 加分项专用 · 直接运行
# ==========================

import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from PIL import Image

# 屏蔽警告
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

# ====================== 路径（自动适配你的 Kaggle） ======================
BASE_PATH = "/kaggle/input/datasets/paultimothymooney/chest-xray-pneumonia/chest_xray"
TRAIN_PATH = os.path.join(BASE_PATH, "train")
TEST_PATH = os.path.join(BASE_PATH, "test")

# ====================== 工具：根据文件名自动打三分类标签 ======================
def get_label(filename):
    if "NORMAL" in filename:
        return 0   # 正常
    elif "virus" in filename:
        return 1  # 病毒肺炎
    elif "bacteria" in filename:
        return 2  # 细菌肺炎
    else:
        return -1

# ====================== 加载图片 + 自动三分类 ======================
def load_data(path):
    images = []
    labels = []
    for folder in ["NORMAL", "PNEUMONIA"]:
        folder_path = os.path.join(path, folder)
        for img_name in os.listdir(folder_path):
            img_path = os.path.join(folder_path, img_name)
            try:
                img = Image.open(img_path).convert("RGB")
                img = img.resize((150, 150))
                img = np.array(img) / 255.0
                label = get_label(img_name)
                if label != -1:
                    images.append(img)
                    labels.append(label)
            except:
                continue
    return np.array(images), np.array(labels)

# 加载数据
print("正在加载训练数据...")
X_train, y_train = load_data(TRAIN_PATH)
print("正在加载测试数据...")
X_test, y_test = load_data(TEST_PATH)

# 独热编码
y_train = to_categorical(y_train, num_classes=3)
y_test = to_categorical(y_test, num_classes=3)

# ====================== 构建三分类模型 ======================
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(150,150,3)),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    
    Conv2D(64, (3,3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    
    Conv2D(128, (3,3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    
    Flatten(),
    Dense(256, activation='relu'),
    Dropout(0.5),
    Dense(3, activation='softmax')  # 3分类
])

model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# ====================== 训练 ======================
history = model.fit(
    X_train, y_train,
    batch_size=32,
    epochs=12,
    validation_split=0.2,
    verbose=1
)

# ====================== 评估 ======================
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_test, axis=1)

acc = accuracy_score(y_true_classes, y_pred_classes)
print("\n===== 三分类测试集准确率 =====")
print(f"Accuracy: {acc:.4f}")

print("\n分类报告：")
print(classification_report(
    y_true_classes, y_pred_classes,
    target_names=["Normal", "Viral", "Bacterial"]
))

# ====================== 混淆矩阵 ======================
cm = confusion_matrix(y_true_classes, y_pred_classes)
plt.figure(figsize=(7,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=["Normal","Viral","Bacterial"],
            yticklabels=["Normal","Viral","Bacterial"])
plt.title('Three-Class Confusion Matrix')
plt.xlabel('Pred')
plt.ylabel('True')
plt.savefig("/kaggle/working/3class_cm.png", dpi=300)
plt.show()

# ====================== 训练曲线 ======================
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.legend()

plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.legend()
plt.savefig("/kaggle/working/3class_curve.png", dpi=300)
plt.show()
```
